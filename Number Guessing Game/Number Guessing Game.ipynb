```python
"""
MINI PROJECT: Number Guessing Game

Objective: Implement a simple number guessiname.

Problem Statement:
Write a Python program to implement a Number Guessing Game where the random mudule generates a random number between 1 and 100.
The program must validate user input against non-integer characters and out-of-range values.
It should restrict the player to a maximum number of attempts (e.g., 10 attempts), provide "Too low" or "Too high" hints,
and end with a Game Over message displaying the secret number if all attempts are completed.p"."
""
```

```python
import random    #import random module to generate secret number

def number_guessing_game():
    secret_number = random.randint(1, 100)
    attempts = 0     #initialize a variable to count valid attempts

    max_attempts = 10
    while attempts < max_attempts:

        #get user input
        guess_number = input("Guess a number(1 to 100):")

        #check input is a number or not
        try:
            guess_number = int(guess_number)
        except ValueError:
            print("Invalid input! Kindly enter numbers only (no letters/symbols)")
            continue    #loop starts again!

        #check for invalid input number
        if guess_number < 1 or guess_number > 100:
            print("You entered an invalid number, kindly enter between 1 and 100.")
        
        else:
        
            attempts += 1     #count valid guesses only

            #compare the guess number with the secret number
            if guess_number < secret_number:
                print("Too low, Try again!")
            elif guess_number > secret_number:
                print("Too high, Try again!")
            else:
                print(f"Congratulations! You guessed a number in {attempts} attempts.")
                break     #stop the game/program when the user guess correctly
            
    else:
        print(f"Game Over! You ran out of attempts.\nThe secret number was {secret_number}.")
```

```python
number_guessing_game()
```
