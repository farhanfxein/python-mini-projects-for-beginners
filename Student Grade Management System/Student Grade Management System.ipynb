```python
"""
MINI PROJECT: Student Grade Management System

Objective: Build a system to store, update, and retrieve student grades.

Problem Statement:
Write a menu-driven Python program to implement a Student Grade Management System using a dictionary.
The system should allow the user to perform the following operations repeatedly until they choose to exit:
1. Add a new student with their grade.
2. Update the grade of an existing student (handle case if student not found).
3. Retrieve/View the grade of a specific student.
Display all student names along with their grades.
Exit the application.
"""
```

```python
#take a dict name students for add students data/info.
students = {}
```

```python
#create a non-parameterized function for adding students details
def add_student():
    name = input("Enter student fullname:")
    grade = input("Enter grade:")
    
    students[name] = grade
    print(f"Student added successfully")
```

```python
#create a non-parameterized function for updating old students grade
def update_grade():
    name = input("Enter student fullname:")
    
    if name in students:
        grade = input("Enter new grade:")
        students[name] = grade
        print(f"Grade updated successfully")
    else:
        print(f"Student not found")
```

```python
#create a non-parameterized function for getting grade of particular student
def get_grade():
    name = input("Enter student fullname: ")
    
    print(f"{name}'s grade: {students.get(name, 'Not found')}")
```

```python
#create a non-parameterized function for showing all students details
def display_all():
    for name, grade in students.items():
        print(f"{name} : {grade}")
```

```python
#using while loop display the description of the choices and do changes as student details requirement.
while True:
    print("\n1. Add Student\n2. Update grade\n3. Get Grade\n4. Display All\n5. Exit")

    #take a choice for specific functions 
    choice = input(f"Enter choice number: ")

    #use ladder method of conditional statements for dealing with choices
    if choice == "1":
        add_student()
    elif choice == "2":
        update_grade()
    elif choice == "3":
        get_grade()
    elif choice ==  "4":
        display_all()
    elif choice == "5":
        print("Exit Successful")     #this line is optional because after exit no details are required.
        break                        #use to exit from the loop
    else:
        print(f"Invalid choice, Try again.")
```
